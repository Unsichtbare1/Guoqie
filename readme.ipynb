work-flow-data

此库训练方法依赖于mortal，因此需要准备好相应的mjai格式牌谱。关于mjai格式请参考mortal。
1. 下载tenhou.net/6格式牌谱。（此格式与mjai格式不同，此类格式是按照牌局的自然发展顺序记录的，能清楚知道摸牌、牌河、鸣牌等信息。而mjai是按照决策来记录的，更加适合训练）
2.  使用tenhou2mjai.py将tenhou.net/6格式牌谱转换为mjai格式牌谱。此脚本会讲输入目录里的tenhou.net/6格式牌谱转换为mjai格式牌谱，并输出到指定目录。
